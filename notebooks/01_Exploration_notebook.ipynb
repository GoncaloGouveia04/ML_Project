# 6. Univariate Analysis

For the sake of analysis we will create a copy of the train model without the 65 empty rows and the impossible values. The train dataset itself is not changed.

```python
candidate = train[~empty_rows].copy()
for column in ["prior_mean_purchase_value", "spend_365d", "days_since_previous_purchase", "prior_mean_gap_days"]:
    impossible = (candidate[column] < 0) | (candidate[column] >= 1e9)
    candidate.loc[impossible, column] = np.nan
```

# 7. Multivariate Analysis

## 7.1 Numeric Variables against the Target

```python
compare = ["days_since_previous_purchase", "prior_mean_gap_days", "prior_purchase_count",
           "purchase_count_90d", "Total a Pagar", "random_noise"]
candidate.groupby(target)[compare].median().T.round(2)
```

We can see that customers who comeback for more pruchases after 90 days, bought more recently and more often than those who don't. The value of the purchases barelly changes and the "random_noise" also doesn't change so they don't seem related to if a client buys again or not.

## 7.2. Correlation with the Target

```python
coorelation_with_target = candidate[numeric + [target]].corr()[target].drop(target).sort_values()

fig, ax = plt.subplots(figsize=(8, 9))
coorelation_with_target.plot(kind="barh", color=PLOT_BLUE, ax=ax)
ax.set(xlabel="correlation with repeat_purchase_90d", title="Which variables move with the target?")
plt.tight_layout()
plt.show()
```

The graph above shows the correlation of the collumns with the target:

- Strongest positive: the purchase counts in the last 90, 180 and 365 days, "spend_180d" and prior_purchase_count. Customers who buy more often, recently and in the past, are more likely to buy again.

- Strongest negative: prior_mean_gap_days, prior_std_gap_days and days_since_previous_purchase. The longer a customer waits between purchases, the smaller the probability of coming back in the next 90 days.

- Close to 0: the money of the current purchase (Total a Pagar, Custo, Total Bruto...) are all close to 0 and random_noise (−0.01) so they don't seem to help us predict if a customer will buy again in the next 90 days.

## 7.3. Correlation between the Numeric Variables

```python
correlations = candidate[numeric].corr()
columns = correlations.columns

for i in range(len(columns)):
    for j in range(i + 1, len(columns)):
        value = correlations.iloc[i, j]
        if abs(value) > 0.9:
            print(columns[i], "and", columns[j], ":", round(value, 2))
```

The collumns in the table above are almost perfectly correlated, so maybe we don't need to keep them all as features for our model.

## 7.4. Categorical Variables against the Target

```python
for column in ["payment_type", "document_series"]:
    rates = candidate.groupby(column)[target].agg(["mean", "count"])
    rates.columns = ["repeat rate", "purchases"]
    display(rates.sort_values("purchases").round(2))
```

In "payment_type" customers who use payment 9 buy again 95% of the time, while other payment types are much lower so it seems related to the target.

In "Document_series" it doesn’t seem as usefull since all three series have the same repetition purchase rate.
